# Introducción a la Herencia

En nuestro viaje por el mundo de la Programación Orientada a Objetos (POO), hemos creado clases que representan entidades en una biblioteca, como `Autor` y `Libro`. Pero, ¿qué pasa si queremos crear nuevas clases que compartan características con las ya existentes, pero con algunas diferencias o adicionales? ¡Aquí es donde entra en juego la herencia!

### Pregunta problema: 
¿Cómo podemos extender nuestras clases existentes para representar diferentes tipos de autores o libros, sin tener que reescribir todo el código?



```python
# Definimos la clase base Autor
class Autor:
    def __init__(self, nombre):
        self.nombre = nombre
```

## Conceptos clave de la herencia

La **herencia** permite que una clase (llamada subclase o clase hija) herede atributos y métodos de otra clase (llamada superclase o clase base). La herencia facilita la reutilización y extensión del código.

- **Herencia simple:** Una clase deriva de una única clase base.
- **Herencia múltiple:** Una clase puede derivar de varias clases base.
- **super():** Función que permite llamar a métodos de la clase base desde la clase derivada.

```python
# Creamos una subclase Escritor que hereda de Autor
class Escritor(Autor):
    def __init__(self, nombre, genero):
        super().__init__(nombre)
        self.genero = genero
```

## Ejemplos prácticos de Herencia

Vamos a extender nuestra clase `Autor` para crear una nueva clase `Escritor` que tenga un atributo adicional: el género literario principal en el que escribe.


```python
# Instanciamos un objeto de la clase Escritor
escritor = Escritor("Mario Benedetti", "Realismo Social")
print(escritor.nombre, escritor.genero)
```

## Herencia Múltiple

En Python, una clase puede heredar de varias clases base, lo que se conoce como herencia múltiple. Aunque puede ser útil, también puede llevar a complicaciones si varias clases base tienen atributos o métodos con el mismo nombre.


```python
# Definimos una segunda clase base
class Academico:
    def __init__(self, universidad):
        self.universidad = universidad

# Creamos una clase que hereda de Escritor y Academico
class EscritorAcademico(Escritor, Academico):
    def __init__(self, nombre, genero, universidad):
        Escritor.__init__(self, nombre, genero)
        Academico.__init__(self, universidad)
```

### Desafío 57: 
Implementa una clase Poeta que herede de Autor y tenga un atributo para el tipo de poesía que escribe.


```python
# Desafío 57: Herencia — clase poeta

# Defino la clase Autor# Crea los atributos nombre y nacionalidad, como privados 
class Autor:
    def __init__(self, nombre, nacionalidad):
        self.__nombre = nombre
        self.__nacionalidad = nacionalidad

    @property
    def nombre(self):
        return self.__nombre

    @property
    def nacionalidad(self):
        return self.__nacionalidad

    @nombre.setter
    def nombre(self, new_nombre):
        if new_nombre != "":
            self.__nombre = new_nombre
        else:
            print("No se pudo modificar el dato")

    @nacionalidad.setter
    def nacionalidad(self, new_nacionalidad):
        if new_nacionalidad != "":
            self.__nacionalidad = new_nacionalidad
        else:
            print("No se pudo modificar el dato")

class Poeta(Autor):
    def __init__(self, nombre, nacionalidad, tipo_poesia):
        # utilizo super().__init__ para heredar los atributos de autor
        super().__init__(nombre, nacionalidad)
        # añado el atributo 
        self.__tipo_poesia = tipo_poesia

    # creo datos para devolver la informacion de cada poeta
    @property
    def datos(self):
        return self.nombre, self.nacionalidad, self.__tipo_poesia

# creo un poeta1
poeta1 = Poeta(
    nombre="Pablo Neruda",
    nacionalidad="Chileno",
    tipo_poesia= "Romántica"
)

# creo un poeta2
poeta2 = Poeta(
    nombre="Federico García Lorca",
    nacionalidad="Español",
    tipo_poesia= "Lírica"
)

# muestro los poetas
print(poeta1.datos)
print(poeta2.datos)
```


### Desafío 58:
Crea una clase Bibliotecario que herede de Usuario y tenga atributos específicos como sección y años_experiencia.


```python
# Definimos una clase Usuario
class Usuario:
    def __init__(self, nombre, apellido):
        self.nombre = nombre
        self.apellido = apellido

# creamos la sublase bibliotecario que hereda de Usuario: nombre y apellido
class Bibliotecario(Usuario):
    def __init__(self, nombre, apellido, seccion, años_experiencia):
        super().__init__(nombre, apellido)
        self.seccion = seccion
        self.años_experiencia = años_experiencia

# creo un bibliotercario
bibliotecario1 = Bibliotecario("Juan", "Pérez", "Literatura", 5)
# lo muestro
print(bibliotecario1.nombre, bibliotecario1.apellido)
print("Seccion:", bibliotecario1.seccion)
print(bibliotecario1.años_experiencia, "años de experiencia.")
```


### Desafío 59:
Diseña una clase LibroDigital que herede de Libro y añada atributos como formato (e.g., PDF, EPUB) y tamaño_archivo. Además, implementa una subclase EBook que sobrescriba un método para mostrar información específica, como enlaces de descarga.


```python
# Desafío 59: LibroDigital - Herencia de clase Libro

# copio libro del desafio_48
class Libro:
    def __init__(self, titulo="", genero="", ISBN=""):
        self.titulo = titulo
        self.genero = genero
        self.ISBN = ISBN

class LibroDigital(Libro):
    def __init__(self, titulo, genero, ISBN, formato, tamaño_archivo):
        # utilizo super().__init__ para heredar los atributos de libro
        super().__init__(titulo, genero, ISBN)
        # añado atributos
        self.__formato = formato
        self.__tamaño_archivo = tamaño_archivo

    # creo datos para devolver la informacion de cada libre digital
    @property
    def mostrar_informacion(self):
        return self.titulo, self.genero, self.ISBN, self.__formato, self.__tamaño_archivo

class EBook(LibroDigital):
    def __init__(self, titulo, genero, ISBN, formato, tamaño_archivo, enlace_descarga):
        # utilizo super().__init__ para heredar los atributos del libro digital
        super().__init__(titulo, genero, ISBN, formato, tamaño_archivo)
        # añado el atributo 
        self.__enlace_descarga = enlace_descarga

    # creo datos para devolver la informacion de cada ebook
    @property
    def mostrar_informacion(self):
        # averigue que se podian heredar funciones tambien
        # puedo aplicarselo a la herencia anterior pero
        # creo que es mas grafico asi...
        return super().mostrar_informacion + (self.__enlace_descarga,)

# creo un LibroDigital
digital1 = LibroDigital(
    titulo="A La Escucha",
    genero="Filosofía",
    ISBN= "9788461090150",
    formato="pdf",
    tamaño_archivo="10.4 MB"
)

# creo un EBook
ebook1 = EBook(
    titulo="A La Escucha",
    genero="Filosofía",
    ISBN= "9788461090150",
    formato="pdf",
    tamaño_archivo="10.4 MB",
    enlace_descarga="https://drive.google.com/drive/u/1/folders/12QbrQKPyCD0sJLB1H7_7LQHHPi7Fm1he"
)

# muestro los datos
print(digital1.mostrar_informacion)
print(ebook1.mostrar_informacion)
```


### Desafío 60:
Implementa una clase EscritorAcademico que herede de Escritor y Academico, e incluya un método adicional para publicar artículos académicos. Asegúrate de utilizar correctamente la función super() para inicializar las clases base.


```python
# Definimos la clase base Autor
class Autor:
    def __init__(self, nombre):
        self.nombre = nombre
        
# Creamos una subclase Escritor que hereda de Autor
class Escritor(Autor):
    def __init__(self, nombre, genero):
        super().__init__(nombre)
        self.genero = genero

# Definimos una segunda clase base
class Academico:
    def __init__(self, universidad):
        self.universidad = universidad

# Creamos una clase que hereda de Escritor y Academico
class EscritorAcademico(Escritor, Academico):
    def __init__(self, nombre, genero, universidad):
        # por lo q entendí, usar Escritor y Academico en herencia multiple es mas practico
        Escritor.__init__(self, nombre, genero)
        Academico.__init__(self, universidad)

    # metodo para agregar un articulo
    def publicar_articulo(self, articulo):
        print(f"Artículo publicado: {articulo}")

escritor1 = EscritorAcademico(
    "Juan Pérez",
    "Ensayo",
    "Universidad de la República"
)

escritor2 = EscritorAcademico(
    "María Rodríguez",
    "Investigación científica",
    "Universidad de Buenos Aires"
)

escritor3 = EscritorAcademico(
    "Carlos Fernández",
    "Divulgación",
    "Universidad de Montevideo"
)

escritor1.publicar_articulo("El impacto de la inteligencia artificial en la educación")
escritor2.publicar_articulo("Nuevas perspectivas sobre el cambio climático")
escritor3.publicar_articulo("Tecnología y desarrollo social en América Latina")
```
