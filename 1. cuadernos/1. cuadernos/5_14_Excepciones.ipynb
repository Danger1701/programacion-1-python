Ejecuta las siguientes líneas y observa que ocurre...

```python
10 * (1/0)
```

```python
4 + spam*3
```

```python
'2' + 2
```

Solicita al usuario dos números enteros e implementa un try-except que maneje la división por cero y los valores no numéricos. Muestra mensajes de error apropiados en cada caso.


```python
# Desafío 67: try-except

# uso el try para que si algo falla, poder trabajar incluso con ese dato (la fasha)
try:
    # ingreso los valores
    num1 = int(input("Ingrese el primer número: "))
    num2 = int(input("Ingrese el segundo número: "))

    # los divido, entendiendo que puede fallar
    resultado = num1 / num2

# si es un problema de tipo de dato ingresado
except ValueError:
    print("Error: debe ingresar valores numéricos.")

# si es un problema de division de un numero entre 0
except ZeroDivisionError:
    print("Error: no se puede dividir por cero.")

# sino hay errores...
# esta figura es opcional, pero me gusto lo que representa en lo didactico,
# ayudando a entender la figura completa del try
else:
    print("Resultado:", resultado)
```


### Desafío 68: 
Crea un programa que tome una lista de valores y realice operaciones matemáticas sobre ellos. Implementa el manejo de varias excepciones comunes como ZeroDivisionError, TypeError, y ValueError.


```python
def random_math(valores):

    try:
        for i in range(len(valores) - 1):
            if i % 2 == 0:
                resultado = float(valores[i]) / float(valores[i + 1])
            else:
                resultado = float(valores[i]) * float(valores[i + 1])
            print(resultado)

    except ZeroDivisionError:
        print("Error: no se puede dividir entre cero.")

    except TypeError:
        print("Error: los valores deben ser numéricos.")

    except ValueError:
        print("Error: valor no válido.")

# 1. ZeroDivisionError
random_math([10, 5, 2, 0, 8])
print("")
# 2. TypeError
random_math([10, 5, None, 2, 8])
print("")
# 3. ValueError
random_math([10, 5, "hola", 2, 8])
```


### Desafío 69: 
Escribe una función que calcule el factorial de un número entero positivo. Maneja las excepciones si el número ingresado es negativo, no es entero, o es demasiado grande para ser procesado.


```python
import math # importo math para calcular factoriales

# creo la excepción propia con class
class NegativeNumberError(Exception):
    pass

class NumberTooBig(Exception):
    pass

# uso el try para que si algo falla, poder trabajar incluso con ese dato (el error)
try:
    # ingreso un numero
    num1 = int(input("Ingrese un número: "))

    # si el numero es negativo
    if num1 < 0:
        # raise lanza manualmente un error, para ser capturado con el except
        raise NegativeNumberError("Error: debe ingresar un valor numerico entero no negativo.")

    # si el numero es ndemasiado grande
    if num1 > 1000:
        # raise lanza manualmente un error, para ser capturado con el except
        raise NumberTooBig("Error: no se puede calcular el factorial de un numero tan grande")

    # calculo su factorial
    resultado = math.factorial(num1)

# si es un problema de tipo de dato ingresado
except ValueError:
    print("Error: debe ingresar un valor numerico entero.")

except NegativeNumberError as error:
    print(error)

except NumberTooBig as error:
    print(error)

else:
    print("Resultado:", resultado)
```


### Desafío 70: 
Crea una excepción personalizada llamada NegativeNumberError que se dispare si el usuario intenta ingresar un número negativo en un programa de cálculo de raíces cuadradas. Implementa el manejo de esta excepción en el programa.


```python
# Desafío 70: excepción personalizada llamada NegativeNumberError

# creo la excepción propia con class
class NegativeNumberError(Exception):
    pass

# uso el try para hacer una ejecucion controlada
try:
    num = int(input("Ingrese un número: "))

    # si el numero es negativo
    if num < 0:
        # raise lanza manualmente un error, para ser capturado con el except
        raise NegativeNumberError("No se puede calcular la raiz cuadrada de un numero negativo")

    resultado = num**(1/2)

# recupera NegativeNumberError
# el as "error", es obligatorio, aunque la variable puede llamarse
except NegativeNumberError as error:
    print(error)

# sino hay errores...
else:
    # simplifico el resultado, sino solo permito 2 digitos
    print(f"resultado: {resultado:.2f}".rstrip("0").rstrip("."))
```
