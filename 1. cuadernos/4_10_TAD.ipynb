# Introducción a los Tipos Abstractos de Datos (TAD)

## Concepto de TAD

Un Tipo Abstracto de Datos (TAD) es una especificación de un conjunto de datos y las operaciones que se pueden realizar con esos datos. El TAD se centra en qué operaciones existen y no en cómo se implementan. Es una herramienta conceptual que nos permite abstraernos de los detalles y pensar en términos de operaciones y datos.

### Ejemplo de TAD: Lista


```python
lista = []

def agregar(elemento):
    lista.append(elemento)

def eliminar(elemento):
    if elemento in lista:
        lista.remove(elemento)

def buscar(elemento):
    return elemento in lista
```

## Desventajas de este enfoque

1. **Falta de encapsulamiento**: En el ejemplo proporcionado, tanto las operaciones como los datos están expuestos y pueden ser modificados directamente desde fuera del TAD. Esto viola el principio de encapsulamiento, que es fundamental para el diseño de software robusto y mantenible.

2. **Dificultad en reutilización**: Con el diseño actual, si quisieras tener múltiples listas y aplicar operaciones sobre ellas, tendrías que crear funciones separadas o modificar las existentes, lo que no es práctico.

3. **Problemas de extensibilidad**: Cambiar la implementación interna (por ejemplo, cambiar de una lista a un árbol binario para mejorar la eficiencia) requeriría modificar todas las funciones que interactúan con la estructura de datos. Esto hace que el código sea difícil de mantener y de extender con nuevas funcionalidades.

4. **Falta de abstracción**: Aunque el TAD debería ayudarnos a abstraernos de los detalles de implementación, el enfoque actual no lo hace de manera efectiva. Esto podría llevar a errores si no se comprenden completamente las implementaciones subyacentes.

### Desafío 44: Sistemas con múltiples entidades interconectadas
Imagina un sistema de gestión de biblioteca que maneja libros, usuarios, préstamos y multas. Usar TADs separados para cada uno de estos elementos podría complicar la interacción y gestión de relaciones entre ellos.


```python
# creo una lista de libros

libros = [
    "Don Quijote de la Mancha",
    "El Principito", 
    "El Señor de los Anillos", 
    "Cien años de soledad",
    ]

# function para agegar un libro
def agregar(libro):
    return libros.append(libro)

# funcion para buscar un libro en la lista
def buscar(libro):
    return libro in libros

# function para eliminar un libro, si existe
def eliminar(libro):
    if libro in libros:
        libros.remove(libro)

# function que muestra todos los libros
def mostrar():
    return print(libros)

# muestro la lista inicial
print("Lista inicial:")
mostrar()

# agrego un libro
agregar("Harry Potter")

# busco un libro que existe y muestro el resultado
print("Existe Harry Potter ?")
print(buscar("Harry Potter"))

# busco un libro que no existe y muestro el resultado
print("Existe Animales Fantásticos ?")
print(buscar("Animales Fantásticos"))

# elimino "Cien años de soledad" de la lista
eliminar("Cien años de soledad")

# elimino un libro q no existe de la lista
eliminar("El Capital")

# muestro la lista final
print("Lista final:")
mostrar()
```


### Desafío 45: Cambio frecuente en requisitos
Supón que estás desarrollando un juego de video con distintos tipos de personajes y armas. Los requerimientos cambian con frecuencia, añadiendo nuevos personajes y habilidades. Mantener y actualizar TADs en este escenario podría ser una tarea titánica.


```python
# creo una lista de libros en español

biblioteca = [
    "Orgullo y prejuicio",
    "El Principito", 
    "El Señor de los Anillos", 
    "Cien años de soledad",
    ]

# creo otra lista de libros en inglés 
library = [
    "Pride and Prejudice",
    "The Little Prince", 
    "The Lord of the Rings", 
    "One Hundred Years of Solitude",
    ]

# function para agegar un elemento a una lista
# primer arg: libro a ingresar, segundo biblioteca
def agregar(elemento, lista):
    return lista.append(elemento)

# funcion para buscar un elemento en una lista
# primer arg: libro a ingresar, segundo biblioteca
def buscar(elemento, lista):
    return elemento in lista

# function para eliminar elemento de lista, si existe
# primer arg: libro a ingresar, segundo biblioteca
def eliminar(elemento, lista):
    if elemento in lista:
        return lista.remove(elemento)

# function que muestra todos los elementos de una lista
# unico argumnento: biblioteca
def mostrar(lista):
    return print(lista)

# muestro la lista inicial
print("Lista inicial de Biblioteca")
mostrar(biblioteca)

# agrego un libro
agregar("Harry Potter", biblioteca)

# busco un libro que existe y muestro el resultado
print("Existe Harry Potter en biblioteca?")
print(buscar("Harry Potter", biblioteca))

# busco un libro que no existe y muestro el resultado
print("Existe Animales Fantásticos en biblioteca?")
print(buscar("Animales Fantásticos", biblioteca))

# elimino "Cien años de soledad" de biblioteca
print("Intento eliminar Cien años de soledad de biblioteca!")
eliminar("Cien años de soledad", biblioteca)

# elimino un libro q no existe de biblioteca
print("Intento eliminar El Capital de biblioteca!")
eliminar("El Capital", biblioteca)

# muestro la lista final
print("Lista final de biblioteca")
mostrar(biblioteca)

# ------

# muestro la lista inicial
print("Lista inicial de library")
mostrar(library)

# agrego un libro
agregar("Harry Potter", library)

# busco un libro que existe y muestro el resultado
print("Existe Harry Potter en library?")
print(buscar("Harry Potter", library))

# busco un libro que no existe y muestro el resultado
print("Existe Fantastic Beast en library?")
print(buscar("Fantastic Beast", library))

# elimino "Cien años de soledad" de library
print("Intento eliminar One Hundred Years of Solitude de library!")
eliminar("One Hundred Years of Solitude", library)

# elimino un libro q no existe de library
print("Intento eliminar Capital de library!")
eliminar("Capital", library)

# muestro la lista final
print("Lista final de library")
mostrar(library)
```
